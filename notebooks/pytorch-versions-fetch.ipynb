```python
def all_nodes(node):
    child_count = 0
    yield node
    if hasattr(node, "children"):
        for child in node.children:
            yield from all_nodes(child)

import pandas as pd
sr = pd.Series(all_nodes(page))
sr
```

```python
def leafs(node):
    child_count = 0
    if hasattr(node, "children"):
        for child in node.children:
            child_count += 1
            yield from leafs(child)
        if child_count == 0:
            yield node
    else:
        yield node

import pandas as pd
leafs_sr = pd.Series(leafs(page))
leafs_sr.name = "bs4"
leafs_sr.index = leafs_sr.apply(id)
df = pd.DataFrame(leafs_sr)
df.index.name="pyid"
df
```

```python
df["type"] = leafs_sr.apply(type).apply(str).str.extract("'(.*)'")
df
```

```python
df["text"] = df.bs4.apply(lambda t: None if t.text.strip() == "" else t.text.strip())
df
```

```python
df = df.dropna()
df.to_csv("pytorch_texts.csv")
df
```
